# 19. Missing Data and Imputation Assumptions

**Prerequisites:** Estimation, conditional probability, and regression. **Estimated time:** 40 minutes.

**Objectives:** Distinguish MCAR, MAR, and MNAR; diagnose observed missingness patterns; select an estimator under explicit assumptions and perform sensitivity analysis.

## Concept recap and assumptions

Missingness mechanisms describe what the probability of being missing depends on. MCAR is independent of data, MAR is explainable by observed variables conditional on them, and MNAR depends on unobserved values even after conditioning. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random
random.seed(19022)

# MAR example: response probability depends on observed group, whose outcome means differ.
rows = []
for group, mean, response_rate in [('A', 10.0, 0.9), ('B', 20.0, 0.3)]:
    for _ in range(1000):
        y = mean + random.gauss(0, 2)
        observed = random.random() < response_rate
        rows.append((group, y if observed else None, response_rate))
full_mean = sum((10.0 if g == 'A' else 20.0) for g, _, _ in rows) / len(rows)
complete = [y for _, y, _ in rows if y is not None]
complete_mean = sum(complete) / len(complete)
# Under the known MAR response model, inverse-probability weighting recovers the target mean.
weighted_sum = sum(y / rate for _, y, rate in rows if y is not None)
weighted_n = sum(1 / rate for _, y, rate in rows if y is not None)
ipw_mean = weighted_sum / weighted_n
print('true target', full_mean, 'complete case', complete_mean, 'IPW', ipw_mean)
assert abs(ipw_mean - full_mean) < abs(complete_mean - full_mean)
assert len(complete) < len(rows)
```

## Comparison and interpretation

Compare complete-case and weighted estimates to the known synthetic target. In real data the target is unknown; report missingness by observed covariates and uncertainty from the imputation/weighting procedure.

## Edge and failure cases

If a response probability is near zero, inverse weights explode and positivity fails. If missingness depends on the missing outcome, this MAR correction is not identified without extra assumptions.
