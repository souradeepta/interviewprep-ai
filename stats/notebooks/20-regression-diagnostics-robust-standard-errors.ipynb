# 20. Regression Diagnostics and Robust Standard Errors

**Prerequisites:** Linear regression, residuals, and sampling variance. **Estimated time:** 40 minutes.

**Objectives:** Use residual diagnostics to identify model problems; compute a heteroskedasticity-consistent slope standard error; distinguish robust inference from model repair.

## Concept recap and assumptions

OLS estimates a conditional mean using a linear projection. Homoskedastic standard errors assume constant residual variance; heteroskedasticity-consistent covariance estimators relax that variance assumption asymptotically, but do not fix a wrong mean function or dependence. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random, math
random.seed(20022)

x = [i / 10 for i in range(1, 101)]
y = [2 + 3 * value + random.gauss(0, 0.02 + 0.04 * value**3) for value in x]
n = len(x); xbar = sum(x) / n; ybar = sum(y) / n
sxx = sum((v - xbar) ** 2 for v in x)
slope = sum((a - xbar) * (b - ybar) for a, b in zip(x, y)) / sxx
intercept = ybar - slope * xbar
resid = [b - (intercept + slope * a) for a, b in zip(x, y)]
sse = sum(e * e for e in resid); sigma2 = sse / (n - 2)
# Homoskedastic textbook SE versus HC0 sandwich slope SE.
naive_se = math.sqrt(sigma2 / sxx)
hc0_se = math.sqrt(sum(((a - xbar) * e) ** 2 for a, e in zip(x, resid)) / (sxx ** 2))
low_var = sum(e*e for a,e in zip(x,resid) if a < 5) / sum(a < 5 for a in x)
high_var = sum(e*e for a,e in zip(x,resid) if a >= 5) / sum(a >= 5 for a in x)
print('slope', slope, 'naive SE', naive_se, 'HC0 SE', hc0_se, 'residual variance low/high', low_var, high_var)
assert high_var > low_var * 5
assert naive_se > 0 and hc0_se > naive_se * 1.08
```

## Comparison and interpretation

Inspect residual-versus-fitted and scale-location plots, leverage/influence, and uncertainty by the dependence unit. Robust standard errors change estimated uncertainty, not OLS coefficients or predictions.

## Edge and failure cases

Small samples, high leverage, serial dependence, or clustered errors can invalidate simple HC0 inference. Use finite-sample corrections, HAC, or cluster-robust methods appropriate to the design.
