# 17. Permutation Tests and Randomization Inference

**Prerequisites:** Hypothesis testing and exchangeability. **Estimated time:** 35 minutes.

**Objectives:** Construct a randomization test under a sharp null; preserve the experimental assignment design; distinguish exact enumeration from Monte Carlo approximation.

## Concept recap and assumptions

A randomization test compares the observed statistic with assignments allowed by the design under a null hypothesis. Permutations are valid only when units are exchangeable under that null; matched and clustered designs require restricted shuffles. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random
random.seed(17022)

# Matched before/after measurements; pair identity is part of the design.
before = [10, 12, 9, 15, 11, 14]
delta = [1, 2, -1, 0, 3, 1]
after = [b + d for b, d in zip(before, delta)]
observed = sum(delta) / len(delta)
# Naive: shuffle labels across all 12 rows, destroying pair membership.
pooled = before + after
naive_stats = []
for _ in range(4000):
    random.shuffle(pooled)
    naive_stats.append(sum(pooled[6:]) / 6 - sum(pooled[:6]) / 6)
# Design-respecting null: flip the sign of each paired difference.
paired_stats = []
for mask in range(1 << len(delta)):
    signed = [d if mask & (1 << i) else -d for i, d in enumerate(delta)]
    paired_stats.append(sum(signed) / len(signed))
p_paired = sum(abs(x) >= abs(observed) for x in paired_stats) / len(paired_stats)
p_naive = (1 + sum(abs(x) >= abs(observed) for x in naive_stats)) / (len(naive_stats) + 1)
print('observed paired mean change', observed, 'naive p', p_naive, 'paired randomization p', p_paired)
assert len(paired_stats) == 64
assert p_paired >= 1 / 64
```

## Comparison and interpretation

Report the observed effect and its randomization p-value; Monte Carlo p-values use a plus-one correction. Randomization inference tests a sharp no-effect null under the actual assignment mechanism.

## Edge and failure cases

A treatment assigned by cluster cannot be shuffled by individual. Highly unbalanced groups and constrained/block randomization also require honoring assignment restrictions.
