# 21. Sequential Experimentation and Peeking Control

**Prerequisites:** A/B testing and multiple testing. **Estimated time:** 40 minutes.

**Objectives:** Explain how repeated fixed-horizon tests inflate false positives; implement a valid simple multiple-look correction; distinguish planned sequential designs from ad hoc stopping.

## Concept recap and assumptions

A fixed-horizon test calibrates its error rate at one planned analysis. Rechecking and stopping at the first significant result changes the procedure. Valid sequential methods use alpha spending, likelihood-ratio boundaries, or always-valid inference designed for optional stopping. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random, math
random.seed(21022)

def two_sided_normal_p(z):
    return math.erfc(abs(z) / math.sqrt(2))
naive_wins = adjusted_wins = 0
trials, looks, alpha = 3000, 15, 0.05
for _ in range(trials):
    observations = []
    naive = adjusted = False
    for look in range(1, looks + 1):
        observations.append(random.gauss(0, 1))  # true null: zero mean
        mean = sum(observations) / len(observations)
        se = 1 / math.sqrt(len(observations))
        p = two_sided_normal_p(mean / se)
        naive |= p < alpha
        # Bonferroni across the 15 planned looks: conservative but valid.
        adjusted |= p < alpha / looks
    naive_wins += naive; adjusted_wins += adjusted
naive_rate = naive_wins / trials; adjusted_rate = adjusted_wins / trials
print('false positive rate, naive looks', naive_rate, 'Bonferroni looks', adjusted_rate)
assert naive_rate > 0.08
assert adjusted_rate < 0.08
```

## Comparison and interpretation

Estimate empirical false-positive rate across repeated null experiments and compare it with the nominal alpha. Bonferroni protects a fixed family of looks but can be conservative; designed sequential procedures can use information more efficiently.

## Edge and failure cases

Changing the number or timing of looks after observing outcomes invalidates the planned correction. Outcome delays and repeated metric selection also create unplanned looks.
