# 16. Bootstrap and Resampling

**Prerequisites:** Probability fundamentals and sampling distributions. **Estimated time:** 35 minutes.

**Objectives:** Construct a nonparametric bootstrap interval; choose the resampling unit; identify settings where bootstrap coverage is unreliable.

## Concept recap and assumptions

The bootstrap approximates an estimator's sampling distribution by sampling observed independent units with replacement. The sample must represent the target population, and dependence must be respected by resampling clusters or blocks. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random
random.seed(16022)

# Positive, right-skewed observations: a normal interval can be implausible for a median.
sample = [0.2, 0.3, 0.4, 0.5, 0.7, 0.9, 1.4, 2.0, 5.8, 9.1]
def median(values):
    values = sorted(values); n = len(values)
    return values[n // 2] if n % 2 else (values[n // 2 - 1] + values[n // 2]) / 2
center = median(sample)
# Naive shortcut: use a normal-theory SE formula for a median.
naive_se = (max(sample) - min(sample)) / (2 * len(sample) ** 0.5)
naive_interval = (center - 1.96 * naive_se, center + 1.96 * naive_se)
# Corrected approach for this lesson: resample independent observations and recompute median.
boot = [median([random.choice(sample) for _ in sample]) for _ in range(3000)]
boot.sort()
bootstrap_interval = (boot[74], boot[2924])
print('median', center, 'naive', naive_interval, 'bootstrap percentile', bootstrap_interval)
assert bootstrap_interval[0] >= min(sample) and bootstrap_interval[1] <= max(sample)
assert len(boot) == 3000
```

## Comparison and interpretation

Compare interval endpoints and width; inspect the bootstrap estimator distribution rather than assuming it is symmetric. A percentile interval is pedagogical here, not universally optimal.

## Edge and failure cases

Resampling one observation produces a degenerate interval. Repeated identical values also make empirical resamples uninformative; dependent observations require cluster/block resampling.

## Exercises

## Summary

Resampling approximates repeated sampling under an empirical distribution. Choosing the right unit and diagnosing small-sample/tail behavior matter more than increasing replicate count.
