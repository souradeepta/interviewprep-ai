## Concept recap and assumptions

Testing many null hypotheses creates opportunities for chance discoveries. Family-wise error controls the probability of any false rejection; false-discovery rate controls the expected false fraction among rejections under stated dependence conditions. This notebook uses only the Python standard library and synthetic data.

## Naive baseline and corrected method

Run the deterministic comparison below. It prints both approaches and asserts key invariants.

```python
import random
random.seed(18022)

def benjamini_hochberg(p_values, q=0.05):
    ordered = sorted(enumerate(p_values), key=lambda pair: pair[1])
    m = len(ordered); cutoff = -1
    for rank, (_, p) in enumerate(ordered, 1):
        if p <= q * rank / m: cutoff = rank
    rejected = {idx for idx, _ in ordered[:cutoff]} if cutoff > 0 else set()
    return rejected
# Synthetic null p-values plus three deliberately small alternatives.
p = [random.random() for _ in range(40)] + [0.0004, 0.001, 0.003]
raw = {i for i, value in enumerate(p) if value < 0.05}
adjusted = benjamini_hochberg(p, q=0.05)
print('raw discoveries', len(raw), 'BH discoveries', sorted(adjusted))
assert len(raw) >= len(adjusted)
assert {40, 41, 42} <= adjusted
```

## Comparison and interpretation

Report the number tested, prespecified family, raw and adjusted discoveries, effect sizes, and uncertainty. BH is a step-up procedure that controls FDR under independence and certain positive dependence structures.

## Edge and failure cases

If many hypotheses are correlated in arbitrary ways, ordinary BH guarantees may not apply. Exploratory selection after seeing results also changes the effective family.
