# Human-Agent Collaboration

Objectives: Human approval, feedback loops, escalation

```
class CollabAgent:
    def __init__(self):
        self.pending = []
    
    def propose(self, decision, confidence):
        if confidence > 0.9:
            self.execute(decision)
        else:
            self.pending.append(decision)
    
    def execute(self, decision):
        print(f'Executed: {decision}')

agent = CollabAgent()
agent.propose('approve order', 0.95)
print('High confidence: auto-executed')
```

**Key Points:** Fundamentals established.

```
class AsyncCollaboration:
    def __init__(self):
        self.pending_review = []
        self.feedback = {}
    
    def agent_proposes(self, task, confidence):
        self.pending_review.append({'task': task, 'conf': confidence})
        return len(self.pending_review) - 1
    
    def human_feedback(self, task_id, approved):
        self.feedback[task_id] = approved

agent = AsyncCollaboration()
task_id = agent.agent_proposes('send email', 0.7)
agent.human_feedback(task_id, True)
print('Async collaboration: non-blocking')
```

**Key Takeaways:** Advanced patterns work.

```
class EscalatingAgent:
    def decide(self, task, confidence, risk):
        if risk == 'high' or confidence < 0.6:
            return 'ESCALATE'
        elif confidence > 0.8:
            return 'AUTO_APPROVE'
        else:
            return 'HUMAN_REVIEW'

agent = EscalatingAgent()
for (t, c, r) in [('a', 0.9, 'low'), ('b', 0.5, 'low'), ('c', 0.95, 'high')]:
    print(f'{t}: {agent.decide(t, c, r)}')
```

**Example 1 Key Points:** Pattern 1 demonstrated.

```
class FeedbackLearner:
    def __init__(self):
        self.decisions = []
    
    def make_decision(self, input_data):
        decision = self._decide(input_data)
        self.decisions.append({'input': input_data, 'decision': decision})
        return decision
    
    def learn_from_feedback(self, decision_id, correct):
        self.decisions[decision_id]['feedback'] = correct
        # Update strategy based on feedback

agent = FeedbackLearner()
id1 = agent.make_decision('data1')
agent.learn_from_feedback(id1, True)
print('Learning from human feedback')
```

```
class TransparentAgent:
    def decide_with_explanation(self, task):
        decision = 'approve'
        explanation = 'Customer has 5-star rating, safe to approve'
        return {'decision': decision, 'why': explanation}

agent = TransparentAgent()
result = agent.decide_with_explanation('order')
print(f'Decision: {result["decision"]} because {result["why"]}')
```

## Baseline implementation and objective

This notebook studies Human Agent Collaboration. The transparent baseline below is a deterministic record validator. Use it to make the control boundary explicit before adding a model or external service. Estimated time: 25 minutes. Objectives: calculate a measurable outcome, compare alternatives, and reason through an edge case. Prerequisites: basic Python and agent workflow concepts.


```
def validate_agent_result(result, required):
    """Return accepted fields and missing requirements without side effects."""
    missing = [name for name in required if not result.get(name)]
    return {"accepted": not missing, "missing": missing}

sample = {"status": "complete", "evidence": "trace-42"}
baseline = validate_agent_result(sample, ["status", "evidence"])
assert baseline["accepted"] is True
assert baseline["missing"] == []
```

## Comparison, metric, and failure cases

Compare the baseline with a permissive check that accepts any nonempty response. Metric: acceptance precision on a labeled set, plus p95 processing latency. A permissive check may be faster but accepts incomplete work. Edge case: an empty evidence string must fail. Failure mode: missing values may be converted to plausible defaults; reject them explicitly and record the reason. Keep raw sensitive values out of diagnostic output.


```
def permissive_check(result):
    return bool(result)

incomplete = {"status": "complete", "evidence": ""}
strict = validate_agent_result(incomplete, ["status", "evidence"])
assert strict["accepted"] is False
assert strict["missing"] == ["evidence"]
assert permissive_check(incomplete) is True  # comparison exposes the unsafe acceptance
```
