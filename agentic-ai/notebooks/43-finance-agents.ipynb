## Concept recap and assumptions

Financial workflow agents must prevent duplicate side effects and reconcile records. This exercise simulates ledger events in memory and never moves money or represents real accounts.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
events = [{'key': 'evt-1', 'amount_cents': 1250}, {'key': 'evt-1', 'amount_cents': 1250}, {'key': 'evt-2', 'amount_cents': -300}]
def apply_events(events):
    seen, balance = set(), 0
    for event in events:
        if event['key'] in seen: continue
        if not isinstance(event['amount_cents'], int): raise ValueError('integer cents required')
        seen.add(event['key']); balance += event['amount_cents']
    return balance, seen
balance, applied = apply_events(events)
print(balance, applied)
assert balance == 950 and len(applied) == 2
```

## 3. Comparison and evaluation

Check duplicate suppression, reconciliation difference, and rejected malformed events. Idempotency keys must be persisted atomically with effects in a real system; this in-memory set does not guarantee that.

## 4. Edge and failure cases

Replay the batch and verify no duplicate is added within one processing pass. Test a missing key, a non-integer amount, and a crash between effect and key persistence as design-level failure cases.

Idempotency prevents duplicate application when events are retried; reconciliation detects disagreement. The example is a local model, not a payment or accounting system.
