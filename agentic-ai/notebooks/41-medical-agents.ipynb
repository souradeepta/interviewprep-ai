## Concept recap and assumptions

This notebook is not a diagnostic system. It uses synthetic urgency labels only to teach conservative routing: uncertain or urgent examples are escalated, never diagnosed.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
cases = [{'id': 'S1', 'signals': ['routine_question'], 'confidence': 0.96}, {'id': 'S2', 'signals': ['urgent_symptom'], 'confidence': 0.99}, {'id': 'S3', 'signals': ['unclear'], 'confidence': 0.42}]
def route(case):
    if 'urgent_symptom' in case['signals']: return 'urgent_human_review'
    if case['confidence'] < 0.8 or 'unclear' in case['signals']: return 'human_review'
    return 'general_information_only'
results = [route(c) for c in cases]
print(results)
assert results == ['general_information_only', 'urgent_human_review', 'human_review']
```

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Track urgent-case sensitivity and abstention coverage, not just accuracy. A high confidence score is not clinical validation and must not bypass the urgent-signal rule.

## 4. Edge and failure cases

Set confidence to 1.0 on an urgent case and verify escalation remains. Missing or malformed signals must route to human review; no output may name a disease or treatment.
