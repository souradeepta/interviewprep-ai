# Customer Service Agents

**Prerequisites:** Structured data and basic Python. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Support automation should minimize retained personal data, cite the supplied policy, and escalate unresolved or sensitive cases. This uses fictional tickets only.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
tickets = [{'id': 'T1', 'topic': 'hours', 'text': 'When are you open?', 'email': 'a@example.test'}, {'id': 'T2', 'topic': 'account_access', 'text': 'I cannot sign in', 'email': 'b@example.test'}]
policy = {'hours': 'Support hours: 09:00–17:00 UTC.'}
def route(ticket):
    if ticket['topic'] not in policy: return {'status': 'escalate', 'ticket_id': ticket['id']}
    return {'status': 'answered', 'text': policy[ticket['topic']], 'source': ticket['topic']}
responses = [route(t) for t in tickets]
print(responses)
assert responses[0]['source'] == 'hours' and responses[1]['status'] == 'escalate'
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Measure supported-answer rate, correct escalation, and unnecessary personal-data exposure. An answer without a policy source is not considered grounded.

## 4. Edge and failure cases

Test a missing topic and a ticket whose text asks the system to ignore policy. Do not place email in the response or logs; prompt text cannot override the routing rules.

## 5. Exercises

Add a policy-backed refund FAQ and a confidence threshold that routes uncertain topics to a person. Assert every automated answer includes a source and every escalation omits email.

## Summary and next practice

Support agents should answer only from approved policy, minimize personal data, and escalate gaps. Synthetic examples do not model real customer cases.
