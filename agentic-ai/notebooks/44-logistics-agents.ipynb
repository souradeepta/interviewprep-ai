# Logistics Agents: Constraints and Dispatch

**Prerequisites:** Python lists and dictionaries. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Dispatch decisions must respect vehicle capacity and delivery windows. The local synthetic example chooses feasible jobs by deadline; it does not create orders or direct real vehicles.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
jobs = [{'id': 'J1', 'weight': 3, 'deadline': 10}, {'id': 'J2', 'weight': 4, 'deadline': 7}, {'id': 'J3', 'weight': 8, 'deadline': 6}]
def dispatch(jobs, capacity):
    chosen, load = [], 0
    for job in sorted(jobs, key=lambda j: (j['deadline'], j['id'])):
        if load + job['weight'] <= capacity:
            chosen.append(job['id']); load += job['weight']
    return chosen, load
route, load = dispatch(jobs, capacity=7)
print(route, load)
assert route == ['J2', 'J1'] and load == 7
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Measure on-time feasibility, capacity utilization, and rejected-job reason. Earliest deadline is a simple heuristic; it is not generally optimal with travel times, geography, or multiple vehicles.

## 4. Edge and failure cases

Reject negative weights and jobs whose individual weight exceeds capacity. A deadline already passed should be surfaced as infeasible, not silently presented as dispatched.

## 5. Exercises

Add an expired job and a validation function for nonnegative weight/deadline. Assert capacity is never exceeded and report each unselected job with a reason.

## Summary and next practice

Operational constraints must be explicit and infeasibility visible. Real dispatch requires live state, route costs, and an authorized human or system boundary.
