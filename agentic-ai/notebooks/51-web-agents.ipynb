# Web Agents

Objectives: Browser automation, API integration, parsing, rate limiting

```
import requests

class WebAgent:
    def fetch(self, url):
        return requests.get(url).text

agent = WebAgent()
html = agent.fetch('https://example.com')
print(f'Fetched {len(html)} bytes')
```

**Key Points:** Fundamentals established.

```
from bs4 import BeautifulSoup

class ParsingAgent:
    def parse(self, html):
        soup = BeautifulSoup(html, 'html.parser')
        links = [a.get('href') for a in soup.find_all('a')]
        return links

agent = ParsingAgent()
links = agent.parse('<html><a href="/page1">Link</a></html>')
print(f'Found {len(links)} links')
```

**Key Takeaways:** Advanced patterns work.

```
import time

class RateLimitedAgent:
    def __init__(self, delay=1):
        self.delay = delay
        self.last_request = 0
    
    def request(self, url):
        elapsed = time.time() - self.last_request
        if elapsed < self.delay:
            time.sleep(self.delay - elapsed)
        self.last_request = time.time()
        return requests.get(url).json()

print('Rate limiting: safe requests')
```

**Example 1 Key Points:** Pattern 1 demonstrated.

```
class ErrorHandlingAgent:
    def fetch_safe(self, url, retries=3):
        for i in range(retries):
            try:
                return requests.get(url, timeout=5).json()
            except Exception as e:
                if i < retries - 1:
                    time.sleep(2 ** i)
        return None

print('Error handling: retries and timeouts')
```

**Example 2 Key Points:** Pattern 2 demonstrated.

```
class HybridAgent:
    def get_data(self, endpoint):
        try:
            return requests.get(f'/api/{endpoint}').json()
        except:
            return self._scrape(endpoint)

print('Hybrid: API first, scrape fallback')
```

## Baseline implementation and objective

This notebook studies Web Agents. The transparent baseline below is a deterministic record validator. Use it to make the control boundary explicit before adding a model or external service. Estimated time: 25 minutes. Objectives: calculate a measurable outcome, compare alternatives, and reason through an edge case. Prerequisites: basic Python and agent workflow concepts.


```
def validate_agent_result(result, required):
    """Return accepted fields and missing requirements without side effects."""
    missing = [name for name in required if not result.get(name)]
    return {"accepted": not missing, "missing": missing}

sample = {"status": "complete", "evidence": "trace-42"}
baseline = validate_agent_result(sample, ["status", "evidence"])
assert baseline["accepted"] is True
assert baseline["missing"] == []
```

## Comparison, metric, and failure cases

Compare the baseline with a permissive check that accepts any nonempty response. Metric: acceptance precision on a labeled set, plus p95 processing latency. A permissive check may be faster but accepts incomplete work. Edge case: an empty evidence string must fail. Failure mode: missing values may be converted to plausible defaults; reject them explicitly and record the reason. Keep raw sensitive values out of diagnostic output.


```
def permissive_check(result):
    return bool(result)

incomplete = {"status": "complete", "evidence": ""}
strict = validate_agent_result(incomplete, ["status", "evidence"])
assert strict["accepted"] is False
assert strict["missing"] == ["evidence"]
assert permissive_check(incomplete) is True  # comparison exposes the unsafe acceptance
```
