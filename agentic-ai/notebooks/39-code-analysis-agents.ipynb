**Prerequisites:** Python basics. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Static analysis must treat source as untrusted data. The agent reports evidence and locations; it never executes inspected code.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
source = 'eval(user_text)\nprint(1 + 1)'
findings = []
for line_no, line in enumerate(source.splitlines(), 1):
    if 'eval(' in line:
        findings.append({'rule': 'dynamic-eval', 'line': line_no, 'severity': 'review'})
print(findings)
assert findings == [{'rule': 'dynamic-eval', 'line': 1, 'severity': 'review'}]
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

The rule is intentionally lexical: comments and safe wrappers can trigger it, while obfuscated calls can evade it. A production analyzer needs parsing, rule provenance, and human review.

## 4. Edge and failure cases

Empty files and a harmless comment containing `eval(` are useful false-positive cases. Include a syntax error and report parse failure separately from zero findings.

## 5. Exercises

Add a rule that flags a hard-coded credential pattern without printing the secret. Test a match and a near miss; preserve line numbers and avoid executing the source.

## Summary and next practice

A finding is evidence for review, not proof of exploitability. Safe analyzers keep the source inert, make rules explainable, and disclose false-positive limits.
