# Research Agents: Evidence and Claim Support

**Prerequisites:** Python lists and dictionaries. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Research assistance should distinguish retrieved evidence from unsupported claims. The corpus is synthetic and supports only a narrow answer; it cannot establish truth beyond these records.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
documents = [{'id': 'D1', 'text': 'The fictional study enrolled 20 participants.'}, {'id': 'D2', 'text': 'The fictional study ran for two weeks.'}]
def answer(claim, docs):
    # Exact phrase retrieval is deliberately conservative and transparent.
    normalized = claim.lower().removeprefix('the ').strip()
    matches = [d for d in docs if normalized in d['text'].lower()]
    if not matches: return {'status': 'unsupported', 'claims': []}
    return {'status': 'needs_verification', 'claims': [{'text': claim, 'sources': [d['id'] for d in matches]}]}
result = answer('The study enrolled 20 participants', documents)
print(result)
assert result['claims'][0]['sources'] == ['D1']
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Measure source coverage and unsupported-claim rate. Keyword overlap is not semantic entailment: source presence does not prove a claim, so this pipeline deliberately marks matches for verification.

## 4. Edge and failure cases

A negated claim can share keywords with a document, while a paraphrase can miss it. Empty retrieval must produce abstention rather than a fluent guess.

## 5. Exercises

Add a claim absent from every source and assert abstention. Add duplicate evidence and de-duplicate source IDs; preserve the document identifiers in the final audit trace.

## Summary and next practice

Research agents should attach provenance and state uncertainty. Retrieval overlap is a candidate evidence signal, never proof that a statement is true.
