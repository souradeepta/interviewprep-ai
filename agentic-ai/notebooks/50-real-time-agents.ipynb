# Real-Time Agents

Objectives: Latency optimization, caching, parallelization, streaming

```
from functools import lru_cache

class CachedAgent:
    @lru_cache(maxsize=1000)
    def cached_decision(self, state):
        # Expensive computation
        return state * 2
    
    def batch_decide(self, states):
        return [self.cached_decision(s) for s in states]

agent = CachedAgent()
results = agent.batch_decide([1, 2, 3])
print(f'Cached decisions: {results}')
```

**Key Points:** Fundamentals established.

```
import asyncio

class ParallelAgent:
    async def decide_async(self, state):
        await asyncio.sleep(0.01)  # Simulate work
        return state
    
    async def parallel_decisions(self, states):
        tasks = [self.decide_async(s) for s in states]
        return await asyncio.gather(*tasks)

# Simulate parallel execution
print('Parallel execution: multiple decisions at once')
```

**Key Takeaways:** Advanced patterns work.

```
class StreamingAgent:
    def stream_results(self, query):
        results = []
        for i in range(3):
            results.append(f'result_{i}')
            yield results  # Return partial results
    
    def get_all_results(self, query):
        all_results = []
        for partial in self.stream_results(query):
            all_results = partial
        return all_results

agent = StreamingAgent()
print('Streaming: return results as available')
```

**Example 1 Key Points:** Pattern 1 demonstrated.

```
import time

class ApproximateAgent:
    def exact_compute(self, data):
        # Slow: 100ms
        time.sleep(0.1)
        return sum(data)
    
    def approximate_compute(self, data):
        # Fast: 1ms
        return sum(data[:10])  # Sample
    
    def smart_decide(self, data, time_budget=0.05):
        if time_budget < 0.01:
            return self.approximate_compute(data)
        else:
            return self.exact_compute(data)

agent = ApproximateAgent()
result = agent.smart_decide([1,2,3,4,5], time_budget=0.001)
print(f'Fast approximation: {result}')
```

**Example 2 Key Points:** Pattern 2 demonstrated.

```
class PrecomputedAgent:
    def __init__(self):
        self.decision_table = self._precompute()
    
    def _precompute(self):
        # Pre-compute all possible decisions
        return {s: s*2 for s in range(100)}
    
    def decide(self, state):
        if state in self.decision_table:
            return self.decision_table[state]  # O(1) lookup
        else:
            return None

agent = PrecomputedAgent()
result = agent.decide(5)
print(f'Pre-computed lookup: {result}')
```

## Baseline implementation and objective

This notebook studies Real Time Agents. The transparent baseline below is a deterministic record validator. Use it to make the control boundary explicit before adding a model or external service. Estimated time: 25 minutes. Objectives: calculate a measurable outcome, compare alternatives, and reason through an edge case. Prerequisites: basic Python and agent workflow concepts.


```
def validate_agent_result(result, required):
    """Return accepted fields and missing requirements without side effects."""
    missing = [name for name in required if not result.get(name)]
    return {"accepted": not missing, "missing": missing}

sample = {"status": "complete", "evidence": "trace-42"}
baseline = validate_agent_result(sample, ["status", "evidence"])
assert baseline["accepted"] is True
assert baseline["missing"] == []
```

## Comparison, metric, and failure cases

Compare the baseline with a permissive check that accepts any nonempty response. Metric: acceptance precision on a labeled set, plus p95 processing latency. A permissive check may be faster but accepts incomplete work. Edge case: an empty evidence string must fail. Failure mode: missing values may be converted to plausible defaults; reject them explicitly and record the reason. Keep raw sensitive values out of diagnostic output.


```
def permissive_check(result):
    return bool(result)

incomplete = {"status": "complete", "evidence": ""}
strict = validate_agent_result(incomplete, ["status", "evidence"])
assert strict["accepted"] is False
assert strict["missing"] == ["evidence"]
assert permissive_check(incomplete) is True  # comparison exposes the unsafe acceptance
```
