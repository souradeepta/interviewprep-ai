# Data Analysis Agents: Query Guardrails

**Prerequisites:** Python lists and simple aggregation. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

An analysis agent needs validated operations and reproducible aggregates. This local table is synthetic; the agent supports a small allowlist and does not execute generated SQL or arbitrary code.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
rows = [{'region': 'north', 'sales': 12}, {'region': 'south', 'sales': 9}, {'region': 'north', 'sales': 7}]
def aggregate(data, group_key, value_key):
    if group_key not in {'region'} or value_key not in {'sales'}: raise ValueError('column not allowed')
    out = {}
    for row in data: out[row[group_key]] = out.get(row[group_key], 0) + row[value_key]
    return out
result = aggregate(rows, 'region', 'sales')
print(result)
assert result == {'north': 19, 'south': 9}
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Check result correctness, query rejection rate, and whether the answer includes grouping and units. A correct aggregate can still mislead if filtering, missingness, or metric definitions are hidden.

## 4. Edge and failure cases

Try a disallowed column and an empty dataset. Decide whether missing values are excluded or treated as zero and make that policy explicit before computing.

## 5. Exercises

Add a row with missing sales and implement an explicit missing-value policy. Tests should cover empty input, rejected field, and exact totals; never use `eval` to interpret a request.

## Summary and next practice

Constrained operations make analyses reproducible and auditable. The example avoids arbitrary query execution and exposes a narrow data contract.
