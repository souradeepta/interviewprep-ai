# Legal Document Agents: Auditability

**Prerequisites:** Text processing and Python dictionaries. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Legal document assistance needs traceable source spans and careful uncertainty handling. The synthetic clause checker flags text for review; it does not interpret law or produce legal conclusions.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
clauses = [{'id': 'C1', 'text': 'Either party may terminate with 30 days written notice.'}, {'id': 'C2', 'text': 'The supplier may change fees at any time.'}]
checks = {'termination': 'terminate', 'unilateral_fee_change': 'change fees at any time'}
def flag(clause):
    hits = [name for name, phrase in checks.items() if phrase in clause['text'].lower()]
    return {'clause_id': clause['id'], 'flags': hits, 'source_text': clause['text'], 'status': 'review' if hits else 'no_rule_match'}
results = [flag(c) for c in clauses]
print(results)
assert results[1]['flags'] == ['unilateral_fee_change'] and results[1]['source_text'] == clauses[1]['text']
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Measure citation completeness and reviewer agreement. A rule match is a retrieval cue, not a determination of enforceability, risk, or applicable law.

## 4. Edge and failure cases

A paraphrase can evade the literal rule; a negated sentence can trigger a false positive. Preserve exact source spans and show no rule match as 'not checked', never as 'safe'.

## 5. Exercises

Add a negated fee clause and ensure the output labels it for review with its source text. Add stable document and clause IDs so an auditor can reproduce each flag.

## Summary and next practice

Auditable assistance preserves provenance and uncertainty. A literal synthetic rule cannot substitute for legal review or jurisdiction-specific analysis.
