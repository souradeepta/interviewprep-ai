## Concept recap and assumptions

Moderation is a policy-routing task with asymmetric errors. These fictional messages are synthetic; the classifier emits review categories and supports human appeal rather than enforcing real policy.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
items = [{'id': 'M1', 'score': 0.93}, {'id': 'M2', 'score': 0.61}, {'id': 'M3', 'score': 0.12}]
def route(item, threshold=0.8):
    if item['score'] >= threshold: return 'human_review_priority'
    if item['score'] >= 0.5: return 'human_review'
    return 'no_flag_from_this_model'
labels = [route(x) for x in items]
print(labels)
assert labels == ['human_review_priority', 'human_review', 'no_flag_from_this_model']
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Evaluate false positives and false negatives by category and subgroup; report review load as the threshold moves. A model score is not a policy judgment and the 'no flag' result is not proof of safety.

## 4. Edge and failure cases

A missing or out-of-range score should be routed to review. Test a borderline score exactly at the threshold and ensure an appeal can reopen a prior decision.
