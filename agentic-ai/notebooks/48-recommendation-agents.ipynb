# Recommendation Agents: Ranking and Consent

**Prerequisites:** Python sorting and dictionaries. **Estimated time:** 30 minutes.

**Objectives:** Identify the domain-specific constraint; implement a bounded deterministic workflow; test failure behavior and explain what the simulation does not establish.

## Concept recap and assumptions

Recommendation systems need relevance, exposure, and user controls. This example ranks fictional catalog items for a synthetic preference profile and applies a consent/availability filter before scoring.

## 1. Transparent local baseline

Run this deterministic, dependency-free baseline. All records are synthetic, and the code has no network access or side effects.

```python
import random
random.seed(20260922)
catalog = [{'id': 'B1', 'tag': 'mystery', 'available': True}, {'id': 'B2', 'tag': 'history', 'available': True}, {'id': 'B3', 'tag': 'mystery', 'available': False}]
def recommend(items, interests, consent):
    if not consent: return []
    eligible = [i for i in items if i['available']]
    return sorted(eligible, key=lambda i: (i['tag'] not in interests, i['id']))
ranked = recommend(catalog, {'mystery'}, consent=True)
print([x['id'] for x in ranked])
assert [x['id'] for x in ranked] == ['B1', 'B2']
assert recommend(catalog, {'mystery'}, consent=False) == []
```

## 2. Extension / realistic constraint

Add one policy or validation boundary and retain a machine-readable reason in the result. Keep the workflow local, deterministic, and reviewable.

## 3. Comparison and evaluation

Track ranking relevance, catalog coverage, and exposure concentration together. This deterministic preference match is not a learned recommender and does not measure user satisfaction.

## 4. Edge and failure cases

Empty interests should fall back to a clearly labeled diverse/popular policy or return no personalized results. Revoked consent must immediately suppress personalization; unavailable items never appear.

## 5. Exercises

Add a deterministic fallback for empty interests that alternates tags. Test consent off, unavailable filtering, stable tie-breaking, and that no personal profile is retained after ranking.

## Summary and next practice

A useful ranker balances relevance with coverage and user controls. Consent and availability checks are hard constraints applied before personalization.
