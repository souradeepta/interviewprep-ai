## Baseline implementation

Inspect how the implementation separates observations from policy. A result should be explainable from its inputs and threshold.


```python
grants={('oncall','read_metrics','aggregate')}
allowed=impl['authorize'](role='oncall',action='read_metrics',resource_class='aggregate',grants=grants)
record=impl['redact']({'model':'v2','email':'user@example.invalid','score':.82})
print(allowed,record)
assert allowed['allowed'] and record['email']=='[REDACTED]'
```
