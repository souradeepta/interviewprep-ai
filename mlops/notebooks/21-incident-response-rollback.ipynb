This notebook uses synthetic local data and a transparent decision policy. Thresholds are examples that must be calibrated to workload and organizational policy; they are not universal production defaults. No cloud credentials or external services are used.

## Baseline implementation

Inspect how the implementation separates observations from policy. A result should be explainable from its inputs and threshold.


```python
incident=impl['assess_incident'](error_rate=.06,latency_p95_ms=620,quality_drop=0,affected_fraction=.3,labels_matured=False)
rollback=impl['should_rollback'](current_error_rate=.06,baseline_error_rate=.005,quality_drop=0,labels_matured=False)
print(incident,rollback)
assert incident['severity']=='SEV1' and rollback['rollback']
```

## Comparison and operational threshold

Change the threshold or input and compare the action. Record the sample floor, policy owner and an action that limits impact.


```python
immature=impl['should_rollback'](current_error_rate=.01,baseline_error_rate=.005,quality_drop=.5,labels_matured=False)
assert not immature['rollback']
```
